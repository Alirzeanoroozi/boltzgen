```python
import torch
import numpy as np

def differs(a, b):
    if type(a) != type(b):
        return True
    if isinstance(a, torch.Tensor):
        if a.shape != b.shape:
            return True
        return not torch.equal(a.cpu(), b.cpu())
    if isinstance(a, np.ndarray):
        if a.shape != b.shape:
            return True
        return not np.array_equal(a, b)
    if isinstance(a, (list, tuple)):
        if len(a) != len(b):
            return True
        return any(differs(x, y) for x, y in zip(a, b))
    return a != b


for k in batch.keys():
    # print(k)
    if k not in masked_batch:
        print(f"Key: {k} is MISSING!")
        continue
    if differs(batch[k], masked_batch[k]):
        print(f"Key: {k} is DIFFERENT!")
        # v = batch[k]
        # print(f"  Type: {type(v)}")
        # try:
        #     print(f"  Shape: {v.shape}")
        # except Exception:
        #     pass
        # print(f"  Value: {v}")
        # print("-" * 100)
```
